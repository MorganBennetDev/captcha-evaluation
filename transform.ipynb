Transform export from label-studio into something we can more easily use.

```python
import json

audio_transformed = []

phonetic_map = {
    "alpha": "a",
    "bravo": "b",
    "charlie": "c",
    "delta": "d",
    "echo": "e",
    "foxtrot": "f",
    "golf": "g",
    "hotel": "h",
    "indian": "i",
    "juliet": "j",
    "kilo": "k",
    "lima": "l",
    "mike": "m",
    "november": "n",
    "oscar": "o",
    "papa": "p",
    "quebec": "q",
    "romeo": "r",
    "sierra": "s",
    "tango": "t",
    "uniform": "u",
    "victor": "v",
    "whiskey": "w",
    "xray": "x",
    "yankee": "y",
    "zulu": "z",
    # I messed up the first half of the annotations for "papa" so we have to do this
    "coffin": "p",
    "zero": "0",
    "one": "1",
    "two": "2",
    "three": "3",
    "four": "4",
    "five": "5",
    "six": "6",
    "seven": "7",
    "eight": "8",
    "nine": "9",
}

def transform_audio_annotation(annotation: str) -> list[str]:
    return [phonetic_map[letter] for letter in annotation.split()]

with open('./captchas/testing/audio.json', 'r') as fp:
    audio = json.load(fp)

    audio_transformed = [{
        "file_name": entry["file_upload"],
        "value": transform_audio_annotation(entry["annotations"][0]["result"][0]["value"]["text"][0])
    } for entry in audio]

with open("./captchas/testing/audio_transformed.json", "w") as fp:
    json.dump(audio_transformed, fp, indent=2, separators=(",", ":"))
```

```python
image_transformed = []

with open('./captchas/testing/image.json', 'r') as fp:
    image = json.load(fp)

    image_transformed = [{
        "file_name": entry["file_upload"],
        "value": [c for c in entry["annotations"][0]["result"][0]["value"]["text"][0]]
    } for entry in image]

with open('./captchas/testing/image_transformed.json', 'w') as fp:
    json.dump(image_transformed, fp, indent=2, separators=(',', ':'))
```
